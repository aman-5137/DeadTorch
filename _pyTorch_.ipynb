## **Tensors**

```python
import torch
print(torch.__version__)
```

```python
# To check if GPU is available to me or not

if torch.cuda.is_available():
    print("GPU is available")
    print("The name of the GPU is", torch.cuda.get_device_name(0))
else:
    print("GPU is not available")


# Click on ram/disk on right to change from CPU to GPU or vice-versa
```

```python
# How to create an empty tensor

x = torch.empty(2,3)
print(x)
```

```python
# What is the type is x?

type(x)
```

```python
# Creating tensor that contains only zeros.

x = torch.zeros(2,3)
x

# NOTE: Here, 2 is columns and 3 is Rows
```

```python
# Same for 1s

x = torch.ones(2,3)
x
```

```python
# How to get random value in torch?

x = torch.rand(2,3)
x
```

```python
# Every time rand runs it generates new value, to control it we can use "manual_seed".

x = torch.manual_seed(10)
x= torch.rand(2,3)
x
```

```python
# How can we create custome tensors?

x = torch.tensor([[1,2,3],[4,5,6]])
x
```

```python
# Other ways to use tensor

# arange function
x = torch.arange(0, 10, 1)
x

# (0 : start, 10: (10-1) end, 2: gap)
```

```python
# To get evenly spaced values in tensor

x = torch.linspace(0, 10, 10)
x
```

```python
#  Identity metrics

x = torch.eye(5)
x
```

```python
# torch.full

x = torch.full((3,3), 5)
x
# (3: column size, 3: row size, 5: element to store)
```
