**Tensor Datatype**

```python
import torch
x = torch.tensor([[1,2,3,4],[5,6,7,8]])
print(x)
x.dtype
```

**Element Wise Operation**
Instead of combining rows and columns like in traditional matrix multiplication, element-wise operations match up elements that share the exact same index location across tensors and compute them one by one.

```python
# Creating new random tensors
a = torch.rand(2,3)
b = torch.rand(2,3)
print(a)
print(b)
```

```python
# Addition
a+b
```

```python
# Subtraction
a - b
```

```python
# Multiplication
a * b
```

```python
# Division(float)

a / b
```

```python
# Division(int)

a//b
```

```python
# Power
a**b
```

```python
# Mod
a % b
```
